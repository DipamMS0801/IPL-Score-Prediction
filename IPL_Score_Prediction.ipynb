```python
#Removing unwanted columns
columns_to_remove = ['mid', 'venue', 'batsman', 'bowler', 'striker', 'non-striker']

print('Before removing unwanted columns: {}'.format(df.shape))
df.drop(labels=columns_to_remove, axis=1, inplace=True)
print('After removing unwanted columns: {}'.format(df.shape))
```

```python
consistent_teams = ['Kolkata Knight Riders', 'Chennai Super Kings', 'Rajasthan Royals',
                    'Mumbai Indians', 'Kings XI Punjab', 'Royal Challengers Bangalore',
                    'Delhi Daredevils', 'Sunrisers Hyderabad']
```

```python
#Keeping only consistent teams
print('Before removing inconsistent teams: {}'.format(df.shape))
df = df[(df['bat_team'].isin(consistent_teams)) & (df['bowl_team'].isin(consistent_teams))]
print('After removing inconsistent teams: {}'.format(df.shape))
```

```python
df['bat_team'].unique()
```

```python
#Removing the first 5 overs data in every match
print('Before removing first 5 overs data: {}'.format(df.shape))
df = df[df['overs']>=5.0]
print('After removing first 5 overs data: {}'.format(df.shape))
```

```python
#Converting the column 'date' from string to datetime object
from datetime import datetime
print("Before converting 'date' column from string to datetime object: {}".format(type(df.iloc[0,0])))
df['date'] = df['date'].apply(lambda x: datetime.strptime(x, '%Y-%m-%d'))
print("After converting 'date' column from string to datetime object: {}".format(type(df.iloc[0,0])))
```

```python
#Selecting correlated features using Heatmap
import matplotlib.pyplot as plt
import seaborn as sns

#Get correlation of all the features of the dataset
corr_matrix = df.corr(numeric_only=True)
top_corr_features = corr_matrix.index

#Plotting the heatmap
plt.figure(figsize=(13,10))
g = sns.heatmap(data=df[top_corr_features].corr(), annot=True, cmap='RdYlGn')
```

```python
#Converting categorical features using OneHotEncoding method
encoded_df = pd.get_dummies(data=df, columns=['bat_team', 'bowl_team'])
encoded_df.columns
```

```python
encoded_df.head(5)
```

```python
# Rearranging the columns
encoded_df = encoded_df[['date', 'bat_team_Chennai Super Kings', 'bat_team_Delhi Daredevils', 'bat_team_Kings XI Punjab',
              'bat_team_Kolkata Knight Riders', 'bat_team_Mumbai Indians', 'bat_team_Rajasthan Royals',
              'bat_team_Royal Challengers Bangalore', 'bat_team_Sunrisers Hyderabad',
              'bowl_team_Chennai Super Kings', 'bowl_team_Delhi Daredevils', 'bowl_team_Kings XI Punjab',
              'bowl_team_Kolkata Knight Riders', 'bowl_team_Mumbai Indians', 'bowl_team_Rajasthan Royals',
              'bowl_team_Royal Challengers Bangalore', 'bowl_team_Sunrisers Hyderabad',
              'overs', 'runs', 'wickets', 'runs_last_5', 'wickets_last_5', 'total']]
```

```python
#Splitting the data into train and test set
X_train = encoded_df.drop(labels='total', axis=1)[encoded_df['date'].dt.year <= 2016]
X_test = encoded_df.drop(labels='total', axis=1)[encoded_df['date'].dt.year >= 2017]

y_train = encoded_df[encoded_df['date'].dt.year <= 2016]['total'].values
y_test = encoded_df[encoded_df['date'].dt.year >= 2017]['total'].values

#Removing the 'date' colum
X_train.drop(labels='date', axis=True, inplace=True)
X_test.drop(labels='date', axis=True, inplace=True)

print("Training set: {} and Test set: {}".format(X_train.shape, X_test.shape))
```

```python
#Linear Regression Model
from sklearn.linear_model import LinearRegression
linear_regressor = LinearRegression()
linear_regressor.fit(X_train, y_train)
```

```python
#Predicting results
y_pred_lr = linear_regressor.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error as mae, mean_squared_error as mse, accuracy_score
print("---- Linear Regression - Model Evaluation -----")
print("Mean Absolute Error (MAE): {}".format(mae(y_test, y_pred_lr)))
print("Mean Squared Error (MSE): {}".format(mse(y_test, y_pred_lr)))
print("Root Mean Squared Error (RMSE): {}".format(np.sqrt(mse(y_test, y_pred_lr))))
```

```python
from sklearn.tree import DecisionTreeRegressor
decision_regressor = DecisionTreeRegressor()
decision_regressor.fit(X_train, y_train)
```

```python
#Predicting results
y_pred_dt = decision_regressor.predict(X_test)
```

```python
# Decision Tree Regression - Model Evaluation
print("---- Decision Tree Regression - Model Evaluation -----")
print("Mean Absolute Error (MAE): {}".format(mae(y_test, y_pred_dt)))
print("Mean Squared Error (MSE): {}".format(mse(y_test, y_pred_dt)))
print("Root Mean Squared Error (RMSE): {}".format(np.sqrt(mse(y_test, y_pred_dt))))
```

```python
#Random Forest Regression Model
from sklearn.ensemble import RandomForestRegressor
random_regressor = RandomForestRegressor()
random_regressor.fit(X_train, y_train)
```

```python
#Predicting results
y_pred_rf = random_regressor.predict(X_test)
```

```python
# Random Forest Regression - Model Evaluation
print("---- Random Forest Regression - Model Evaluation ----")
print("Mean Absolute Error (MAE): {}".format(mae(y_test, y_pred_rf)))
print("Mean Squared Error (MSE): {}".format(mse(y_test, y_pred_rf)))
print("Root Mean Squared Error (RMSE): {}".format(np.sqrt(mse(y_test, y_pred_rf))))
```

```python
#Predictions

def predict_score(batting_team='Chennai Super Kings', bowling_team='Mumbai Indians', overs=5.1, runs=50, wickets=0, runs_in_prev_5=50, wickets_in_prev_5=0):
  temp_array = list()

  # Batting Team
  if batting_team == 'Chennai Super Kings':
    temp_array = temp_array + [1,0,0,0,0,0,0,0]
  elif batting_team == 'Delhi Daredevils':
    temp_array = temp_array + [0,1,0,0,0,0,0,0]
  elif batting_team == 'Kings XI Punjab':
    temp_array = temp_array + [0,0,1,0,0,0,0,0]
  elif batting_team == 'Kolkata Knight Riders':
    temp_array = temp_array + [0,0,0,1,0,0,0,0]
  elif batting_team == 'Mumbai Indians':
    temp_array = temp_array + [0,0,0,0,1,0,0,0]
  elif batting_team == 'Rajasthan Royals':
    temp_array = temp_array + [0,0,0,0,0,1,0,0]
  elif batting_team == 'Royal Challengers Bangalore':
    temp_array = temp_array + [0,0,0,0,0,0,1,0]
  elif batting_team == 'Sunrisers Hyderabad':
    temp_array = temp_array + [0,0,0,0,0,0,0,1]

  # Bowling Team
  if bowling_team == 'Chennai Super Kings':
    temp_array = temp_array + [1,0,0,0,0,0,0,0]
  elif bowling_team == 'Delhi Daredevils':
    temp_array = temp_array + [0,1,0,0,0,0,0,0]
  elif bowling_team == 'Kings XI Punjab':
    temp_array = temp_array + [0,0,1,0,0,0,0,0]
  elif bowling_team == 'Kolkata Knight Riders':
    temp_array = temp_array + [0,0,0,1,0,0,0,0]
  elif bowling_team == 'Mumbai Indians':
    temp_array = temp_array + [0,0,0,0,1,0,0,0]
  elif bowling_team == 'Rajasthan Royals':
    temp_array = temp_array + [0,0,0,0,0,1,0,0]
  elif bowling_team == 'Royal Challengers Bangalore':
    temp_array = temp_array + [0,0,0,0,0,0,1,0]
  elif bowling_team == 'Sunrisers Hyderabad':
    temp_array = temp_array + [0,0,0,0,0,0,0,1]

  # Overs, Runs, Wickets, Runs_in_prev_5, Wickets_in_prev_5
  temp_array = temp_array + [overs, runs, wickets, runs_in_prev_5, wickets_in_prev_5]

  # Converting into numpy array
  temp_array = np.array([temp_array])

  # Prediction
  return int(linear_regressor.predict(temp_array)[0])
```

```python
# ============================================
# IPL FINAL SCORE PREDICTION SYSTEM
# ============================================

teams = [
    'Chennai Super Kings',
    'Delhi Daredevils',
    'Kings XI Punjab',
    'Kolkata Knight Riders',
    'Mumbai Indians',
    'Rajasthan Royals',
    'Royal Challengers Bangalore',
    'Sunrisers Hyderabad'
]


# --------------------------------------------
# Select Team
# --------------------------------------------

def select_team(team_type):

    print(f"\nSelect {team_type}:")

    for i, team in enumerate(teams, start=1):
        print(f"{i}. {team}")

    while True:
        try:
            choice = int(input(f"Enter {team_type} number (1-8): "))

            if 1 <= choice <= 8:
                return teams[choice - 1]
            else:
                print("Please enter a number between 1 and 8.")

        except ValueError:
            print("Please enter a valid number.")


# --------------------------------------------
# Make Prediction
# --------------------------------------------

def make_prediction():

    print("\n" + "=" * 50)
    print("       IPL FINAL SCORE PREDICTION")
    print("=" * 50)

    # Batting Team
    batting_team = select_team("Batting Team")

    # Bowling Team
    while True:

        bowling_team = select_team("Bowling Team")

        if bowling_team != batting_team:
            break

        print("\nBatting and Bowling teams cannot be the same!")
        print("Please select another bowling team.")

    # ----------------------------------------
    # Overs
    # ----------------------------------------

    while True:

        try:

            overs = float(
                input("\nEnter overs completed (e.g. 9.2): ")
            )

            whole_overs = int(overs)
            balls = round((overs - whole_overs) * 10)

            if whole_overs < 0 or whole_overs > 20:
                print("Overs must be between 0 and 20.")

            elif balls < 0 or balls > 5:
                print(
                    "Invalid overs!\n"
                    "Use cricket notation such as 5.2, 9.4 or 12.5."
                )

            elif whole_overs == 20 and balls > 0:
                print("A T20 innings cannot exceed 20 overs.")

            else:
                break

        except ValueError:
            print("Please enter a valid number.")

    # ----------------------------------------
    # Current Runs
    # ----------------------------------------

    while True:

        try:

            runs = int(input("Enter current runs: "))

            if runs >= 0:
                break

            print("Runs cannot be negative.")

        except ValueError:
            print("Please enter a valid integer.")

    # ----------------------------------------
    # Wickets
    # ----------------------------------------

    while True:

        try:

            wickets = int(
                input("Enter wickets lost (0-10): ")
            )

            if 0 <= wickets <= 10:
                break

            print("Wickets must be between 0 and 10.")

        except ValueError:
            print("Please enter a valid integer.")

    # ----------------------------------------
    # Runs in Previous 5 Overs
    # ----------------------------------------

    while True:

        try:

            runs_in_prev_5 = int(
                input("Enter runs scored in previous 5 overs: ")
            )

            if runs_in_prev_5 >= 0:
                break

            print("Runs cannot be negative.")

        except ValueError:
            print("Please enter a valid integer.")

    # ----------------------------------------
    # Wickets in Previous 5 Overs
    # ----------------------------------------

    while True:

        try:

            wickets_in_prev_5 = int(
                input("Enter wickets lost in previous 5 overs: ")
            )

            if 0 <= wickets_in_prev_5 <= 10:
                break

            print("Wickets must be between 0 and 10.")

        except ValueError:
            print("Please enter a valid integer.")

    # ----------------------------------------
    # Prediction
    # ----------------------------------------

    final_score = predict_score(
        batting_team=batting_team,
        bowling_team=bowling_team,
        overs=overs,
        runs=runs,
        wickets=wickets,
        runs_in_prev_5=runs_in_prev_5,
        wickets_in_prev_5=wickets_in_prev_5
    )

    # ----------------------------------------
    # Display Result
    # ----------------------------------------

    print("\n" + "=" * 50)
    print("             PREDICTION RESULT")
    print("=" * 50)

    print(f"Batting Team           : {batting_team}")
    print(f"Bowling Team           : {bowling_team}")
    print(f"Overs                  : {overs}")
    print(f"Current Runs           : {runs}")
    print(f"Wickets                : {wickets}")
    print(f"Runs in Previous 5     : {runs_in_prev_5}")
    print(f"Wickets in Previous 5  : {wickets_in_prev_5}")

    print("\nPredicted Final Score Range:")
    print(
        f"{final_score - 10} to {final_score + 5}"
    )

    print("=" * 50)


# ============================================
# MAIN MENU
# ============================================

while True:

    print("\n" + "=" * 50)
    print("       IPL SCORE PREDICTION SYSTEM")
    print("=" * 50)

    print("1. Predict Score")
    print("2. Exit")

    choice = input("\nEnter your choice (1/2): ").strip()

    if choice == "1":

        make_prediction()

        # Predict Again
        while True:

            again = input(
                "\nDo you want to predict another match? (Y/N): "
            ).strip().lower()

            if again == "y":
                break

            elif again == "n":

                print("\n" + "=" * 50)
                print("Thank you for using IPL Score Prediction System!")
                print("Goodbye!")
                print("=" * 50)

                break

            else:
                print("Please enter Y or N.")

        # Exit after user chooses N
        if again == "n":
            break

    elif choice == "2":

        print("\n" + "=" * 50)
        print("Thank you for using IPL Score Prediction System!")
        print("Goodbye!")
        print("=" * 50)

        break

    else:

        print("\nInvalid choice! Please enter 1 or 2.")
```
